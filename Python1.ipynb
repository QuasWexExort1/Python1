```python
matrix = (
    (5, 1, 2, 3),
    (4, 0, 1, 6),
    (3, 2, 0, 1)
)

sums = [sum(row[i] for row in matrix) for i in range(len(matrix[0]))]

min_sum = sums[0] + sums[1]
best_cols = (0, 1)

for i in range(1, len(sums) - 1):
    current_sum = sums[i] + sums[i+1]
    if current_sum < min_sum:
        min_sum = current_sum
        best_cols = (i, i + 1)

print(best_cols)
```

```python
numbers = [1, 2, 3, 4, 5]
k = 2
C = 99

numbers.append(0)

for i in range(len(numbers) - 1, k, -1):
    numbers[i] = numbers[i - 1]

numbers[k] = C
print(numbers)
```

```python
def superset(a, b):
    if a == b:
        print("Множества равны")
    elif a > b:
        print(f"Объект {a} является чистым супермножеством")
    elif b > a:
        print(f"Объект {b} является чистым супермножеством")
    else:
        print("Супермножество не обнаружено")

superset({1, 2, 3}, {1, 2})
```

```python
cats = [
    ('Мартин', 5, 'Алексей', 'Егоров'),
    ('Фродо', 3, 'Анна', 'Самохина'),
    ('Вася', 4, 'Андрей', 'Белов')
]

cats_dict = {}
for cat in cats:
    cats_dict[cat[0]] = {
        'age': cat[1],
        'owner_name': cat[2],
        'owner_surname': cat[3]
    }

print(cats_dict)
```
